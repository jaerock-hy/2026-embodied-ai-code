## MSE and Cross Entropy functions

```python
import numpy as np
import random
```

```python
# mse
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

```python
# mae
def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))
```

```python
# test mse function with random data
y_true = np.array([1, 2, 3])
y1_pred = np.array([1, 2, 2.2])
y2_pred = np.array([10, 2, 3])
print(mse(y_true, y1_pred))  # should print 0.21333333333333334
print(mse(y_true, y2_pred))  # should print 27.0
```

```python
# test mae function with random data
y_true = np.array([1, 2, 3])
y1_pred = np.array([1, 2, 2.2])
y2_pred = np.array([10, 2, 3])
print(mae(y_true, y1_pred))  # should be a more less than 0.2133
print(mae(y_true, y2_pred))  # should be a lot less than 27.0
```

```python
# cross-entropy
def cross_entropy_org(y_true, y_pred):
    return -np.sum(y_true * np.log(y_pred))
```

```python
def cross_entropy(y_true, y_pred):
    # add a small value to y_pred to avoid log(0)
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    return -np.sum(y_true * np.log(y_pred))
```

```python
# test the cross_entropy_org function with y_pred 
y_true = np.array([1, 2, 0])
y_pred = np.array([0.8, 0.1, 0.2])
print(cross_entropy_org(y_true, y_pred))  # should print a large value due to log(0) being undefined
```

```python
# test the cross_entropy_org function with y_pred containing a zero value to see how it handles log(0)
y_true = np.array([1, 2, 0])
y_pred = np.array([0.8, 0, 0.2])
print(cross_entropy_org(y_true, y_pred))  # should print a large value due to log(0) being undefined
```

```python
# test the cross_entropy_org function with y_pred containing a zero value to see how it handles log(0)
y_true = np.array([1, 2, 0])
y_pred = np.array([0.8, 0, 0.2])
print(cross_entropy(y_true, y_pred))  # should print a large value due to log(0) being undefined
```

### Cross Entropy for Mini-Batch

```python
# cross entropy for mini-batch 
def cross_entropy(y_true, y_pred):
    batch_size = 1 if y_true.shape[0] == 1 else y_true.shape[0]

    # add a small value to y_pred to avoid log(0)
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    return -np.sum(y_true * np.log(y_pred)) / batch_size
```
